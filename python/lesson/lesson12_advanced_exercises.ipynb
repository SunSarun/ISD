## Exercise 1: Robust numeric parser

- Write a function that parses user-provided string to int or float.
- Raise/handle exceptions for invalid formats.
- Return tuple `(success, value_or_error)` and print examples.

```python
def parse_number(s):
    try:
        return True, int(s)
    except ValueError:
        try:
            return True, float(s)
        except ValueError:
            return False, f"Could not parse '{s}' as int or float"

samples = ["42", "3.14", "-10", "abc", "12.34.56"]
for item in samples:
    success, result = parse_number(item)
    print(f"Input: {item!r} -> Success: {success}, Value: {result}")
```

## Exercise 2: Multi-exception workflow

- Build a division function that converts both inputs with `float()` and divides, using `try/except/else/finally`.
- Handle `TypeError`, `ValueError`, and `ZeroDivisionError` separately.
- Demonstrate all branches with different calls.

```python
def safe_divide(a, b):
    try:
        num_a = float(a)
        num_b = float(b)
        result = num_a / num_b
    except TypeError:
        print("TypeError: both inputs must be valid data types")
    except ValueError:
        print("ValueError: string could not be converted to float")
    except ZeroDivisionError:
        print("ZeroDivisionError: division by zero is not allowed")
    else:
        print(f"Result: {result}")
    finally:
        print("Operation complete")

safe_divide(10, 2)
safe_divide("10", "0")
safe_divide("10", "abc")
safe_divide(None, 5)
```

## Exercise 3: None-aware processing

- Given list with values and `None`, compute average of valid numbers only.
- Ignore `None` safely.
- Print cleaned list and average.

```python
raw_values = [10, None, 20, 30, None, 40, None, 50]
cleaned = [x for x in raw_values if x is not None]
average = sum(cleaned) / len(cleaned) if cleaned else 0
print(f"Cleaned list: {cleaned}")
print(f"Average: {average}")
```

## Exercise 4: Input validation loop

- Prompt repeatedly for integer between 1 and 100 until valid.
- Use exception handling and custom error messages.
- Print accepted value and attempt count.

```python
attempts = 0
while True:
    attempts += 1
    user_input = input("Enter an integer between 1 and 100: ")
    try:
        val = int(user_input)
        if 1 <= val <= 100:
            print(f"Accepted value: {val}")
            print(f"Total attempts: {attempts}")
            break
        else:
            print("Number must be between 1 and 100.")
    except ValueError:
        print("Invalid input. Please enter a valid integer.")
```

## Exercise 5: Mini CLI app

- Create text menu app (add/subtract/multiply/divide/quit).
- Use input + try/except for invalid choices and values.
- Keep looping until quit command.

```python
while True:
    print("1. Add")
    print("2. Subtract")
    print("3. Multiply")
    print("4. Divide")
    print("5. Quit")
    choice = input("Choose an option (1-5): ").strip()
    if choice == "5" or choice.lower() == "quit":
        print("Exiting application.")
        break
    if choice not in ("1", "2", "3", "4"):
        print("Invalid option selected.")
        continue
    try:
        a = float(input("Enter first number: "))
        b = float(input("Enter second number: "))
        if choice == "1":
            print(f"Result: {a + b}")
        elif choice == "2":
            print(f"Result: {a - b}")
        elif choice == "3":
            print(f"Result: {a * b}")
        elif choice == "4":
            try:
                print(f"Result: {a / b}")
            except ZeroDivisionError:
                print("Error: Cannot divide by zero.")
    except ValueError:
        print("Error: Invalid numeric input.")
```
