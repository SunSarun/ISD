## Exercise 1: Array statistics

- Create numeric array/list with at least 12 elements.
- Compute min, max, sum, average, and median (manual or built-in helper).
- Print all metrics.

```python
import statistics

numbers = [12, 45, 7, 23, 56, 89, 34, 67, 90, 11, 44, 78]

min_val = min(numbers)
max_val = max(numbers)
sum_val = sum(numbers)
avg_val = sum_val / len(numbers)
med_val = statistics.median(numbers)

print(f"Min: {min_val}")
print(f"Max: {max_val}")
print(f"Sum: {sum_val}")
print(f"Average: {avg_val:.2f}")
print(f"Median: {med_val}")
```

## Exercise 2: Iterator protocol deep dive

- Create iterator from a string and consume values with `next()` until exhausted.
- Handle `StopIteration` explicitly.
- Print total number of consumed characters.

```python
text = "PythonIterator"
it = iter(text)
count = 0

while True:
    try:
        char = next(it)
        print(char, end=" ")
        count += 1
    except StopIteration:
        print()
        break

print(f"Total consumed characters: {count}")
```

## Exercise 3: Custom iterator class

- Implement class `EvenNumbers(limit)` yielding even numbers up to limit.
- Use `__iter__` and `__next__`.
- Iterate and print values.

```python
class EvenNumbers:
    def __init__(self, limit):
        self.limit = limit
        self.current = 2

    def __iter__(self):
        return self

    def __next__(self):
        if self.current <= self.limit:
            val = self.current
            self.current += 2
            return val
        raise StopIteration

for num in EvenNumbers(20):
    print(num)
```

## Exercise 4: Iterator state with `next()`

- Create a custom iterator class `Countdown(start)` that counts down to 1.
- Use `iter()` and `next()` to take the first two values manually.
- Then use a `for` loop on the same iterator to get the remaining values.
- Add a short comment explaining why the `for` loop continues where `next()` stopped.

```python
class Countdown:
    def __init__(self, start):
        self.current = start

    def __iter__(self):
        return self

    def __next__(self):
        if self.current >= 1:
            val = self.current
            self.current -= 1
            return val
        raise StopIteration

cd = iter(Countdown(10))
print(next(cd))
print(next(cd))
for val in cd:
    print(val)
```
